```python
import pandas as pd
import matplotlib.pyplot as plt

# ─── PARAMETERS ───────────────────────────────────────────
date      = "2018-05-21"   # ← change this date
data_type = "load"         # ← "load", "wind", or "solar"
zone      = "Coast"        # ← change this zone/asset
# ──────────────────────────────────────────────────────────

path = f"/Users/val/Desktop/Princeton/{data_type}_{zone}_{date}.csv"
df = pd.read_csv(path, index_col=0, header=[0,1])

data = df.iloc[1:, :24].astype(float)
hours = range(24)

plt.figure(figsize=(12, 5))

for i in range(len(data)):
    plt.plot(hours, data.iloc[i].values, color='steelblue', alpha=0.02)

plt.plot(hours, data.mean().values, color='red', linewidth=2, label='Mean')

plt.title(f"{data_type.capitalize()} scenarios - {zone} - {date}")
plt.xlabel("Hour")
plt.ylabel("MW")
plt.xticks(hours)
plt.legend()
plt.tight_layout()
plt.show()
```
